```python
#Creating a clean dataframe
#Variables to keep zqinc (Annual Household Income), bq9_1 (Party Identification)
#and zq101 (Ethnicity)
clean_df = df[["zqinc", "bq9_1", "zq101"]].copy()

#Renaming variables for clarity
clean_df.columns = ["annual_income", "political_party", "ethnicity"]
```

```python
#Cleaning the data for the target minorities
target_minorities = [7,8,9,11,12]
clean_df = clean_df[clean_df["ethnicity"].isin(target_minorities)]
```

```python
#Cleaning the dataset
clean_df["annual_income"] = clean_df["annual_income"].replace([-2,-1,17], np.nan)
clean_df["political_party"] = clean_df["political_party"].replace([-2,-1,15], np.nan)
```

```python
clean_df.info()
```

```python
clean_df = clean_df.dropna()
```

```python
clean_df.info()
```

#### RQ Is Annual Household Income associated with Minority’s Political Party Identification?

H0 (Null Hypothesis): There is no significant association between the annual household income and political party identification among minorities. Annual household income and minority's political party identifications are independent.

H1 (Alternative Hypothesis): There is a significant association between the annual household income and political party identification among minorities. Annual household income and minority political party identifications are independent.

```python
# Preparing the dataset
income_vs_pparty = pd.crosstab(clean_df["annual_income"], clean_df["political_party"])

# Statistical test (Chi-square test)
chi2, p_value, dof, expected = chi2_contingency(income_vs_pparty)

print(f'\n---- Statistical test ---')
print(f'Chi-square statistic: {chi2:.3f}')
print(f'p-value: {p_value:.5f}')

# Interpretation 
alpha = 0.05
if p_value < alpha:
    print("Conclusion: Reject the Null Hypothesis (Statistically Significant association has been found.)")
else:
    print("Conclusion: Fail to Reject the Null Hypothesis (No statistically significant association has been found.)")
```
